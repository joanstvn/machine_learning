## 2. Codifique variables categóricas usando LabelEncoder y One-Hot Encoding

En Machine Learning, los algoritmos matemáticos requieren entradas numéricas:
- **LabelEncoder**: Asigna a cada categoría única un número entero consecutivo ($0, 1, 2, \dots$). Se recomienda para variables ordinales o variables objetivo.
- **One-Hot Encoding**: Genera una columna binaria ($0$ o $1$) por cada categoría. Es ideal para variables nominales sin orden jerárquico intrínseco (evita que el modelo asuma una falsa relación de magnitud).

```python
%pip install scikit-learn imbalanced-learn
```

```python
from sklearn.preprocessing import LabelEncoder

# Crear una copia del DataFrame para el preprocesamiento
df_prep = df.copy()

# 2.1. Aplicar LabelEncoder a la variable categórica 'categoria'
label_encoder_cat = LabelEncoder()
df_prep['categoria_encoded'] = label_encoder_cat.fit_transform(df_prep['categoria'])

# Mostrar el mapeo de categorías a valores numéricos
mapeo_categoria = dict(zip(label_encoder_cat.classes_, label_encoder_cat.transform(label_encoder_cat.classes_)))
print("Mapeo generado por LabelEncoder ('categoria'):")
for cat, code in mapeo_categoria.items():
    print(f"  {cat} -> {code}")

df_prep[['categoria', 'categoria_encoded']].head()
```

```python
# 2.2. Aplicar One-Hot Encoding a la variable nominal 'ciudad'
df_prep = pd.get_dummies(df_prep, columns=['ciudad'], prefix='ciudad', dtype=int)

# Columnas generadas tras One-Hot Encoding
columnas_ciudad = [col for col in df_prep.columns if col.startswith('ciudad_')]
print("Columnas One-Hot generadas para 'ciudad':", columnas_ciudad)
df_prep[['producto'] + columnas_ciudad].head()
```

## 3. Aplique normalización y estandarización

El escalado previene que variables numéricas con rangos muy amplios dominen indebidamente el aprendizaje del modelo:
- **Normalización (Min-Max Scaling)**: Escala los datos a un rango definido, comúnmente $[0, 1]$:
  $$X_{norm} = \frac{X - X_{min}}{X_{max} - X_{min}}$$
- **Estandarización (StandardScaler / Z-Score)**: Centra los datos con media $\mu = 0$ y desviación estándar $\sigma = 1$:
  $$Z = \frac{X - \mu}{\sigma}$$

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler

# 3.1. Normalización (MinMaxScaler) en 'cantidad' y 'precio_unitario'
scaler_minmax = MinMaxScaler()
df_prep[['cantidad_norm', 'precio_unitario_norm']] = scaler_minmax.fit_transform(
    df_prep[['cantidad', 'precio_unitario']]
)

# 3.2. Estandarización (StandardScaler) en 'total_venta'
scaler_std = StandardScaler()
df_prep['total_venta_std'] = scaler_std.fit_transform(df_prep[['total_venta']])

# Comparación de columnas originales vs escaladas
df_prep[[
    'cantidad', 'cantidad_norm',
    'precio_unitario', 'precio_unitario_norm',
    'total_venta', 'total_venta_std'
]].head()
```

## 4. Balancee la variable objetivo usando SMOTE

**SMOTE (Synthetic Minority Over-sampling Technique)** es una técnica de sobremuestreo para datasets desbalanceados. En lugar de duplicar registros, genera nuevas instancias sintéticas de la clase minoritaria mediante interpolación geométrica de vecinos más cercanos ($k$-NN).

Definiremos un objetivo de clasificación: clasificar si una transacción corresponde a una **'Venta Alta'** ($1$, clase minoritaria: top 20% en monto) o una **'Venta Regular'** ($0$, clase mayoritaria: 80%).

```python
from imblearn.over_sampling import SMOTE
from collections import Counter

# 4.1. Definir la variable objetivo binaria 'venta_alta'
umbral_alta = df_prep['total_venta'].quantile(0.80)
df_prep['venta_alta'] = (df_prep['total_venta'] >= umbral_alta).astype(int)

# 4.2. Definir la matriz de características (X) y el vector objetivo (y)
feature_cols = ['cantidad_norm', 'precio_unitario_norm', 'categoria_encoded'] + columnas_ciudad
X = df_prep[feature_cols]
y = df_prep['venta_alta']

print("Distribución de clases ANTES de SMOTE:")
conteo_antes = Counter(y)
for clase, cantidad in conteo_antes.items():
    print(f"  Clase {clase}: {cantidad} muestras ({cantidad / len(y):.1%})")
```

```python
# 4.3. Aplicar SMOTE para balancear las clases equitativamente
smote = SMOTE(random_state=42)
X_resampled, y_resampled = smote.fit_resample(X, y)

print("Distribución de clases DESPUÉS de SMOTE:")
conteo_despues = Counter(y_resampled)
for clase, cantidad in conteo_despues.items():
    print(f"  Clase {clase}: {cantidad} muestras ({cantidad / len(y_resampled):.1%})")

print(f"\nFilas originales: {len(y)} | Filas balanceadas: {len(y_resampled)}")
```

## 5. Muestre el conjunto de datos final listo para entrenar un modelo de clasificación

Con todas las transformaciones completadas, verificamos la estructura final y realizamos la división canónica en conjunto de **Entrenamiento (80%)** y **Prueba (20%)** mediante `train_test_split` estratificado.

```python
from sklearn.model_selection import train_test_split

# 5.1. Partición de entrenamiento y prueba
X_train, X_test, y_train, y_test = train_test_split(
    X_resampled,
    y_resampled,
    test_size=0.20,
    random_state=42,
    stratify=y_resampled
)

print("=== Resumen de Conjuntos para Entrenamiento de Clasificación ===")
print(f"X_train (Entrenamiento): {X_train.shape[0]} registros, {X_train.shape[1]} características")
print(f"X_test  (Prueba):        {X_test.shape[0]} registros, {X_test.shape[1]} características")
print(f"y_train distribución:    {dict(Counter(y_train))}")
print(f"y_test  distribución:    {dict(Counter(y_test))}")
```

```python
# 5.2. Vista final del dataset preparado y balanceado
df_final = pd.DataFrame(X_resampled, columns=feature_cols)
df_final['venta_alta'] = y_resampled

print(f"Dimensiones del dataset final listo para clasificar: {df_final.shape}")
print("\nPrimeras 5 filas:")
df_final.head()
```
